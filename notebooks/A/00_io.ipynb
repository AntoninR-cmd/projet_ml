## Afficher le nombre de spectres, de colonnes, de blocs et de lignes dans le premier bloc

Un **bloc** (*row group*) est un paquet de lignes consécutives que parquet stocke et compresse séparément, colonne par colonne. On peut donc lire un seul bloc, et seulement certaines de ses colonnes, sans charger le reste du fichier.

```python
print(f"Nombre de spectre : {pf.metadata.num_rows}")
print(f"Nombre de colonnes : {pf.metadata.num_columns}")
print(f"Nombre de blocs : {pf.metadata.num_row_groups}")
print(f"Nombre de lignes par bloc : {pf.metadata.row_group(0).num_rows}")
```
